# Task 1: Generate Student Dataset using NumPy

```python
import numpy as np
```

```python
# Generate data for 60 students:

Student_id=[]
for i in range(1001,1061):
    Student_id.append(i)
    
subject=['Math', 'programing', 'state', 'english']
Marks=np.random.randint(0,100,(60,4))


print("Student _ides:", Student_id)
print("Marks Matrix:", Marks)
print("shape of Marks:", Marks.shape)
print("type of marks", Marks.dtype)

  
```

# Task 2: Basic Statistical Analysis

```python
# average marks of each Subject
math_marks=Marks[:, :1]
math_total_marks=0
for i in math_marks:
    math_total_marks+=i
math_average_marks=math_total_marks/60
print ("Mathe Average Marks:", math_average_marks)
programe_marks=Marks[:, 1:2]
programe_total_marks=0
for i in programe_marks:
    programe_total_marks+=i
programe_average_marks=programe_total_marks/60
print ("programing Average Marks:", programe_average_marks)
state_marks=Marks[:, 2:3]
state_total_marks=0
for i in state_marks:
    state_total_marks+=i
state_average_marks=state_total_marks/60
print ("state Average Marks:", state_average_marks)
eng_marks=Marks[:, 3:4]
eng_total_marks=0
for i in eng_marks:
    eng_total_marks+=i
eng_average_marks=eng_total_marks/60
print ("english Average Marks:", eng_average_marks)

# overall class average

class_average=math_average_marks+programe_average_marks+state_average_marks+eng_average_marks
class_average=class_average/4
print("class average:", class_average)

subject_std=np.std(Marks, axis=0)
print('standardevation of each subjects:')
for i in range(4):
    print(subject[i], ":", subject_std[i])
print("Sunject Maximum marks:", np.max(Marks, axis=0))
print("sunject minimum marks:", np.min(Marks, axis=0))
```

# Task 3: Student Performance Report

```python

import numpy as np

# Calculate total marks of each student
total_marks = np.sum(Marks, axis=1)

print("Total Marks of Each Student:")
print(total_marks)

# Define categories
excellent = total_marks >= 320
good = (total_marks >= 250) & (total_marks <= 319)
average = (total_marks >= 200) & (total_marks <= 249)
poor = total_marks < 200

# Count students in each category
print("\nStudents in Each Category:")
print("Excellent:", np.sum(excellent))
print("Good:", np.sum(good))
print("Average:", np.sum(average))
print("Poor:", np.sum(poor))

# Student IDs of Excellent students
print("\nExcellent Students IDs:")
print(np.array(Student_id)[excellent])

# Student IDs of Poor students
print("\nPoor Students IDs:")
print(np.array(Student_id)[poor])
```

# Task 4: Find Top Students

```python
# total marks of each student
total_marks=np.sum(Marks, axis=1)
# find top indes
top5_index=np.argsort(total_marks)[-5:][::-1]
for i in top5_index:
    print("Rank:", i)
    print("student_id", Student_id[i])
    print("total marks:", total_marks[i])
    print("subject wise:")
    for j in range(4):
        print(subject[j],':', Marks[i, j])
        print("-"*30)
```

# Task 6: Identify Weak Students

```python

import numpy as np

# Find marks below 40
below_marks = Marks < 40

# Count subjects below 40 for each student
weak_count = np.sum(below_marks, axis=1)

# Students weak in at least 2 subjects
weak_student = weak_count >= 2

print("Weak Student IDs:")
print(np.array(Student_id)[weak_student])

print("Weak Student Marks:")
print(Marks[weak_student])

print("Subject-wise weak student details:")

# Get indexes of weak students
for i in np.where(weak_student)[0]:
    print("\nStudent ID:", Student_id[i])

    for j in range(4):
        print(subject[j], ":", Marks[i, j])

    print("-" * 30)
```

# Task 7: Reshape + Aggregation

```python
# Task 8
# Functions
def student_report(Students_id, marks):
    print("Student ID:", Students_id)
    for i in range(4):
        print(subject[i],":",Marks[i])
    total = sum(marks)
    percentage = (total / 400) * 100
    # Grade
    if percentage >= 85:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"
    print("Total:", total,"/ 400")
    print("Percentage:", percentage,"%")
    print("Grade:", grade)
    print("--------------------------------------------------------" )
# display All records
for i in range(60):
    student_report(Student_id[i], Marks[i])
# user input
user_input=int(input('Enter number 1001-1060'))
index=user_input-1001
if (index<0):
    print("Invalid: index")
elif (index>=60):
    print("Invalid: index")
else:
    student_report(Student_id[index], Marks[index])
```
