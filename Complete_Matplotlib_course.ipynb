# Basics of Matplotlib 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_axis=np.linspace(0,5,11)
x_axis
```

```python
y_axis=x_axis**2
y_axis
```

### plot() function that are used plot the graph line graph

```python
plt.plot(x_axis, y_axis)
plt.title("hello")
plt.xlabel("X_axis")
plt.ylabel("X_axis")
```

## Subplots in matplotlib

### subplot(row, column, plotno) function that are usde draw multple plot in one figr 

```python
plt.subplot(2,2,1)
plt.plot(x_axis, y_axis)
plt.subplot(2,2,2)
plt.plot(y_axis, x_axis)
plt.subplot(2,2,3)
plt.plot(x_axis, y_axis)
plt.subplot(2,2,4)
plt.plot(x_axis**2, y_axis**2)
plt.tight_layout()
```

## object oriented way

### two function that used to create manual plot or graph 
### 1 fiqure()  2 add_axis(left, botom, width, height) these are paramenters that help to create the plot 
### fiqure(figsize=(6,6), dpi=100 defualt set by on we set 300 or 400 )

```python
fig=plt.figure(figsize=(9,9), dpi=300)
axis_1=fig.add_axes([0.1, 0.1, 0.8,0.8])
plt.plot(x_axis, y_axis)
axis_2=fig.add_axes([0.3,0.3,0.5,0.5])
plt.plot(x_axis, y_axis)
```

## Types of plots 
### these are three types of plots 
### 1_ scatter(x,y) 2_ hist(x)  3_ boxplot(x)

```python
plt.scatter(x_axis, y_axis)
```

```python
plt.hist(x_axis)
```

```python
plt.boxplot(x_axis)
```

## Saving plots
### savefig("image or file name") 

```python
fig=plt.figure()
ax=fig.add_axes([0.1, 0.1, 0.8,0.8])
plt.plot(x_axis, y_axis)
plt.savefig("plot.png")
```
